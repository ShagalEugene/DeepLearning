````python
best_model = None

learning_rates = [1e-3, 5e-3]
regularization_strengths = [1e-3, 1e-2, 1e-1]
hidden_dims = [80, 100]
weight_scales = [1e-2, 5e-2]
best_val_acc = -1
for hd in hidden_dims:
    for lr in learning_rates:
        for reg in regularization_strengths:
            for ws in weight_scales:                
                model = TwoLayerNet(
                    hidden_dim=hd,
                    input_dim=input_size,
                    num_classes=10,
                    weight_scale=ws,
                    reg=reg,
                )
                
                solver = Solver(
                    model, data,
                    update_rule='sgd',
                    optim_config={
                        'learning_rate': lr,
                    },
                    lr_decay=0.95,
                    num_epochs=10,
                    batch_size=100,
                    verbose=False
                )
                solver.train()

                val_acc = solver.check_accuracy(data['X_val'], data['y_val'])

                print(f'hd: {hd}; lr: {lr}; reg: {reg}; ws: {ws}; val_acc: {val_acc:.3f}')
                
                if best_val_acc < val_acc:
                    best_val_acc = val_acc
                    best_model = model
````

1) При обучении модели на большем наборе данных, у неё улучшится обобщающая способность (будут выявляться характерные закономерности для классов). В целом модели будет сложно переобучиться на большем объёме.

2) Если увеличим число нейронов, то модель станет сложнее, а в нашем случае это ухудшит ситуацию (так как модель и так сложная для данных, так как она их запоминает).

3) Регуляризаця штрафует модель за большие веса (l2 заставляет модель распределять вклад по многим признакам, а не опираться на конкретные) или за использование большого количества весов (l1 штрафует сумму модулей весов и зануляет часть из них, делая модель разреженной).
